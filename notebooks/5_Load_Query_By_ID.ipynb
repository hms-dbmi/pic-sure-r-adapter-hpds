# 5. Load a saved query by ID

`picsure::loadQueryByID(session, id)` fetches the body of a previously-saved query and returns it as a clause / clause-group handle. From there you can hand it to `runQuery()` (shown below), pass it to one of the export functions (`exportAsPFB`, `exportCSV`, `exportTSV`), or nest it inside a new `buildClauseGroup()` group to refine the cohort.

If you only need to run a saved query without inspecting or modifying it, see notebook 6 (`runQueryByID`) for the one-step shortcut.

```r
library(picsure)
```

```r
token_file <- "token.txt"
my_token <- readLines(token_file, warn = FALSE)[1]
```

```r
session <- picsure::connect(
  platform = picsure::Platform$BDC_AUTHORIZED,
  token    = my_token
)
```

Replace the UUID below with one of your own saved-query IDs (copy from the PIC-SURE web UI).

```r
clause_group <- picsure::loadQueryByID(session, "ea5f32d2-2ca8-4820-a96a-41239af75d38")
picsure::runQuery(session, clause_group)
```
