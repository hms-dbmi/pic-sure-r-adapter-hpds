# 6. Run a saved query by ID in one step

`picsure::runQueryByID(session, id, type = "count")` is a convenience wrapper that loads a saved query and immediately runs it. It's equivalent to `runQuery(session, loadQueryByID(session, id))` — same result type as `runQuery()`.

Use this when you don't need to inspect or modify the loaded clause group before execution.

```r
library(picsure)
```

```r
token_file <- "token.txt"
my_token <- readLines(token_file, warn = FALSE)[1]
```

```r
session <- picsure::connect(
  platform = picsure::Platform$BDC_AUTHORIZED,
  token    = my_token
)
```

Replace the UUID below with one of your own saved-query IDs from the PIC-SURE web UI.

```r
picsure::runQueryByID(session, "ea5f32d2-2ca8-4820-a96a-41239af75d38")
```
