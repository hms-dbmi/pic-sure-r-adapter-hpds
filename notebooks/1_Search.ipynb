# 1. Search the dictionary

`picsure::searchDictionary(session, term)` returns matching dictionary entries as a `data.frame`. Narrow a search to a subset of studies or curated subgroups with a **FacetSet** built via `picsure::facets(session)` and `picsure::addFacet()`.

```r
library(picsure)
```

```r
token_file <- "token.txt"
my_token <- readLines(token_file, warn = FALSE)[1]
```

```r
authorized_session <- picsure::connect(
  platform = picsure::Platform$BDC_AUTHORIZED,
  token    = my_token
)
```

## Plain search

Returns a `data.frame` with one row per matching variable. Columns include `name`, `display`, `conceptPath`, `values`, and study/dataset metadata.

```r
picsure::searchDictionary(authorized_session, "age")
```

## Search inside a curated subgroup

`addFacet()` accepts a scalar or vector of values under the same key. The returned FacetSet is the same handle passed in, mutated in place; you can chain calls.

```r
child_facet <- picsure::facets(authorized_session)
picsure::addFacet(child_facet, "Consortium_Curated_Facets", "06m-post index")

picsure::searchDictionary(authorized_session, "", facets = child_facet)
```

## Filter to a single study (`dataset_id`)

```r
study_facet <- picsure::facets(authorized_session)
picsure::addFacet(study_facet, "dataset_id", "phs000007")
```

```r
picsure::searchDictionary(authorized_session, "sex", facets = study_facet)
```

## Lighter payloads

Pass `include_values = FALSE` to omit each variable's enumerated values from the response — useful when you only need names and descriptions.

```r
picsure::searchDictionary(authorized_session, "age", include_values = FALSE)
```
