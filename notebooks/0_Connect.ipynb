# 0. Connect

Open a session against a PIC-SURE platform. `picsure::connect()` takes a `Platform` enum member (or its label) and a personal access token from your PIC-SURE web UI's **User Profile** tab.

On the first call of an R session, `reticulate` provisions an isolated Python environment via `uv` — this can take a few seconds the first time. Subsequent calls reuse the cached environment.

```r
library(picsure)
```

```r
token_file <- "token.txt"
my_token <- readLines(token_file, warn = FALSE)[1]
```

```r
authorized_session <- picsure::connect(
  platform = picsure::Platform$BDC_AUTHORIZED,
  token    = my_token
)
```

```r
open_session <- picsure::connect(
  platform = picsure::Platform$BDC_DEV_OPEN,
  token    = my_token
)
```

```r
nhanes_open <- picsure::connect(
  platform = picsure::Platform$NHANES_OPEN,
  token    = my_token
)
```
